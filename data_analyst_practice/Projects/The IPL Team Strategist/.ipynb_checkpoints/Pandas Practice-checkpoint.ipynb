## Batting Runs Scored by each team till now

```python
# 
print(deliveries.groupby("batting_team")['batsman_runs'].sum().sort_values(ascending=False))
```

#### Batting Runs Scored by each player in a  team till now

```python
print(deliveries.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10))
```

# Strike_rate

```python
runs=deliveries.groupby('batter')['batsman_runs'].sum()
balls=deliveries.groupby('batter').size()
strike_rate=(runs / balls ) *100

strike_rate[runs > 1000].sort_values(ascending=False).head(10)
```

```python
runs=deliveries.groupby('batter')['batsman_runs'].sum()
balls=deliveries.groupby('batter').size()
strike_rate=(runs / balls ) * 100

fair_sr=strike_rate[ balls >= 500 ]
fair_sr.sort_values(ascending=False).head(10)
```
