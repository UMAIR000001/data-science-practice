# LEVEL 3: Grouping & Aggregation

- Calculate total runs scored by each batter.

- Calculate total balls faced by each batter.

- Calculate total runs scored by each team.

- Calculate strike rate for each batter (total_runs / balls_faced * 100).

- Use .agg() to get both total runs and total balls faced per batter in one line.

- Use .reset_index() after your groupby to convert index to column.

- Find top 5 batters by total runs.

- Find top 5 batters by strike rate, but only include those who faced ≥ 100 balls.

```python
deliveries.groupby('batter')["batsman_runs"].sum().sort_values(ascending=False).reset_index()
```

```python
deliveries.groupby('batter').size().sort_values(ascending=False).reset_index().rename(columns={0:"total_balls_faced"})
```

```python
deliveries.groupby('batting_team')["batsman_runs"].sum().sort_values(ascending=False).reset_index().head(10).rename(columns={"batsman_runs":"total_runs"})
```

```python
#Calculate strike rate for each batter (total_runs / balls_faced * 100).
total_runs=deliveries.groupby('batter')["batsman_runs"].sum()
balls_faced=deliveries.groupby('batter').size()
(total_runs / balls_faced * 100).sort_values(ascending=False).reset_index().rename(columns={0:"strike_rate"})
```

```python
deliveries.groupby("batter").agg(
    total_runs=('batsman_runs','sum'),
    total_balls=('batsman_runs','size')
).reset_index()
```

```python
#Calculate strike rate for each batter (total_runs / balls_faced * 100) with  filter batters with ≥ 500 balls
total_runs=deliveries.groupby('batter')["batsman_runs"].sum()

balls_faced=deliveries.groupby('batter').size()

strike_rate=(total_runs / balls_faced * 100)
#strike_rate
fair_strike_rate=strike_rate[balls_faced>500].sort_values(ascending=False).reset_index()
fair_strike_rate.head()
```
