# Aggregation:
- Aggregate Funtions=Reduces a set of values into a single Summary Value
- It is used to Summarize and Analyze data
  

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Umair\OneDrive\Desktop\data-science-practice\data_analyst_practice\Python\Pandas\data.csv",index_col="No")
df
```

```python
 
#print(df.mean())  It will through Errors as this dataset also consist of columns of having text values
print(df.mean(numeric_only=True))
```

```python
print("The Mean Values at each Column:")
print(df.mean(numeric_only=True))
print("\n")
print("The Sum Values at each Column:")
print(df.sum(numeric_only=True))
print("\n")
print("The Minimum Values at each Column:")
print(df.min(numeric_only=True))
print("\n")
print("The Maximum Values at each Column:")
print(df.max(numeric_only=True))
print("\n")
print("The Total Count values  of Each  Column:")
print(df.count())
print("\n")
```

### Now for Specifc Column

```python
print("The Mean Values at each Column:")
print(df["Height"].mean())

print("The Sum Values at each Column:")
print(df["Height"].sum())

print("The Minimum Values at each Column:")
print(df["Height"].min())

print("The Maximum Values at each Column:")
print(df["Height"].max())

print("The Total Count values  of Each  Column:")
print(df["Height"].count())

```

### Using Aggreagate Funtions With groupby()
- Aggreate Funtions are used to Summarize and Analyse Data, often used with  the groupby() Funtion

```python
Group=df.groupby("Type1")
print("The Mean Value of Height Column:")
print(Group["Height"].mean())

print("The Sum Values Height Column:")
print(Group["Height"].sum())

print("The Minimum Values Height Column:")
print(Group["Height"].min())

print("The Maximum Values Height Column:")
print(Group["Height"].max())

print("The Total Count values Height Column:")
print(Group["Height"].count())

```

# Data Cleaning 

##### Data Cleaning =The Process of Fixing/Removing,Incomplete,incorrect or irrelevant data
#####  75% of Work Done With Pandas is data Cleaning

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Umair\OneDrive\Desktop\data-science-practice\data_analyst_practice\Python\Pandas\data.csv",index_col="No")
df
```

```python
### Steps :
Drop Irrelevant Columns
Handle Missing Data
Fix Inconsistent Values
Standardize text
Fix Data types
Remove Duplicate Values
```

```python
# Drop Irrelevant Columns:
df=df.drop(columns=["Legendary"])
df
```

```python
# Drop Multiple Irrelevant Columns:
df=df.drop(columns=["Height","Weight"])
df
```

```python
#2. Handle Missing data(i.e NAN)
df=df.dropna(subset=["Type1","Type2"])
print(df.to_string())
```

```python
# 3. Fix Incosistent values 
df["Type1"]=df["Type1"].replace({"Water":"WATER","Grass":"GRASS"})
print(df.to_string())
```

```python
# 4. Standardize Text :
df["Name"]=df["Name"].str.lower()
df["Type1"]=df["Type1"].str.lower()
df["Type2"]=df["Type2"].str.lower()


print(df.to_string())
```

```python
# 5. Fix Data Types
df["Legendary"]=df["Legendary"].astype(bool)
print(df.to_string())
```

```python
# 6. Remove Duplicate Values:
df=df.drop_duplicates()
print(df.to_string())
```
