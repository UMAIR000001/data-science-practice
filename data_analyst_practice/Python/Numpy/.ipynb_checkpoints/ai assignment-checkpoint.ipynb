```python
# --- IMPORTS ---
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# ==========================================
# PART A: CLASSIFICATION MODELS
# ==========================================
print("--- PART A RESULTS ---")

# 1. Load Data
iris_data = load_iris()
features = iris_data.data
labels = iris_data.target

# 2. Split Data (80% Training, 20% Testing)
# We use random_state=10 to get a different shuffle than others
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2, random_state=10)

# --- MODEL 1: KNN (Using k=5) ---
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
knn_preds = knn_model.predict(X_test)

print("\n1. K-Nearest Neighbors (k=5)")
print(f"Accuracy: {accuracy_score(y_test, knn_preds):.2f}")
print("First 10 Predictions:", knn_preds[:10])
print(classification_report(y_test, knn_preds))

# --- MODEL 2: SVM ---
svm_model = SVC(kernel='linear')
svm_model.fit(X_train, y_train)
svm_preds = svm_model.predict(X_test)

print("\n2. Support Vector Machine (SVM)")
print(f"Accuracy: {accuracy_score(y_test, svm_preds):.2f}")
print(classification_report(y_test, svm_preds))

# --- MODEL 3: Perceptron ---
perc_model = Perceptron()
perc_model.fit(X_train, y_train)
perc_preds = perc_model.predict(X_test)

print("\n3. Perceptron")
print(f"Accuracy: {accuracy_score(y_test, perc_preds):.2f}")
print(classification_report(y_test, perc_preds))




```

```python
# ==========================================
# PART C: CROSS VALIDATION
# ==========================================
print("\n--- PART C RESULTS ---")
# Testing the SVM model 5 times
cv_results = cross_val_score(svm_model, features, labels, cv=5)
print("Fold Accuracies:", cv_results)
print(f"Average Accuracy: {cv_results.mean():.2f}")
print("Why CV? It gives a more reliable estimate of model performance.")
```

```python
# ==========================================
# PART B: OVERFITTING & REGULARIZATION
# ==========================================
print("\n--- PART B RESULTS ---")

# Overfitted Model (k=1 is too specific)
knn_bad = KNeighborsClassifier(n_neighbors=1)
knn_bad.fit(X_train, y_train)
train_score_bad = knn_bad.score(X_train, y_train)
test_score_bad = knn_bad.score(X_test, y_test)

# Regularized Model (k=50 is very smooth)
knn_good = KNeighborsClassifier(n_neighbors=50)
knn_good.fit(X_train, y_train)
train_score_good = knn_good.score(X_train, y_train)
test_score_good = knn_good.score(X_test, y_test)

print(f"Overfitted (k=1)   -> Train: {train_score_bad:.2f}, Test: {test_score_bad:.2f}")
print(f"Regularized (k=50) -> Train: {train_score_good:.2f}, Test: {test_score_good:.2f}")

print("\nExplanation: Overfitting is when the model learns the noise in the training set instead of the actual pattern.")
```

```python
# ==========================================
# PART D: VISUALIZATIONS
# ==========================================
print("\n--- PART D GRAPHS ---")

# Graph 1: Scatter Plot
# Using 'coolwarm' palette to look different
df_vis = pd.DataFrame(features, columns=iris_data.feature_names)
df_vis['Species'] = labels

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_vis, x='sepal length (cm)', y='sepal width (cm)', hue='Species', palette='coolwarm')
plt.title("Iris Species Distribution")
plt.show()

# Graph 2: Confusion Matrix (For SVM)
cm = confusion_matrix(y_test, svm_preds)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, cmap='Reds') # Changed to Red color
plt.title("SVM Confusion Matrix")
plt.ylabel("True Label")
plt.xlabel("Predicted Label")
plt.show()

# Graph 3: Model Comparison
models = ['KNN', 'SVM', 'Perceptron']
scores = [accuracy_score(y_test, knn_preds), accuracy_score(y_test, svm_preds), accuracy_score(y_test, perc_preds)]

plt.bar(models, scores, color=['purple', 'orange', 'gray']) # Different colors
plt.title("Accuracy Comparison")
plt.ylim(0.5, 1.05)
plt.show()
```
