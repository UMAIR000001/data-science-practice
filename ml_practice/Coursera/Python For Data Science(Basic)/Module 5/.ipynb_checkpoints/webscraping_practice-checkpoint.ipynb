# 🌟 HTML Revision Notes (Anime Edition)
### 1. HTML Basics

- HTML is like the world of One Piece 🌊 → it gives the structure to everything, just like how the Grand Line connects all islands.

- Tags are like characters’ roles → <p> is like Luffy’s laugh (paragraph), is like Levi’s serious command (heading).

### 2. Table Tags 🍽️

- Think of a table like a Blue Lock football field:

- <table> → The whole football ground.

- <tr> (table row) → A line of players standing.

- <td> (table data) → Each player in the line.

- <th> (table header) → The coach’s instructions written at the top.

#### ⚽ Example:
Imagine Isagi, Bachira, and Chigiri in one row → that’s <tr>. Each one is a <td>.

### 3. Rows <tr>

- Rows are like Attack on Titan walls 🧱 → they are stacked horizontally one after another.


- Example: Wall Maria, Wall Rose, Wall Sheena → each one = <tr>.

### 4. Columns <td>

- Inside each row, you put columns.

- Like in Solo Leveling, Sung Jin-Woo summons his shadows. Each shadow (Igris, Beru, Iron) is one <td> inside his squad row <tr>.

## HTML Demo (Anime Page)

```python
%%html
<html>
  <body>
    <h1>Anime Power Levels</h1>

    <table>
      <tr><td>Luffy</td><td>Gear 5</td></tr>
    </table>

    <table>
      <tr><td>Asta</td><td>Demon Form</td></tr>
    </table>
  </body>
</html>
```

### ✅ What’s happening:
We are writing a small HTML page that has two tables: one for Luffy, one for Asta.
Think of this as the "Treasure Map" 🗺️ — we’re creating the world before we explore it.

## Store HTML into Python Variable

```python

# Imagine this is a page from "Anime Character Database"
html_doc = """
<html>
  <body>
    <h1>Anime Power Levels</h1>

    <table>
      <tr><td>Luffy</td><td>Gear 5</td></tr>
    </table>

    <table>
      <tr><td>Asta</td><td>Demon Form</td></tr>
    </table>
  </body>
</html>
"""

```

####  ✅ What’s happening:
We copied the HTML into a Python string (html_doc).
This is like putting the treasure map into your backpack 🎒 so Luffy can carry it on the adventure.

## Parse with BeautifulSoup + Search Tables

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html_doc, "html.parser")

# find() → like Zoro spotting the FIRST enemy he sees
first_table = soup.find("table")
print("find() result:", first_table)

# find_all() → like Luffy scanning the WHOLE battlefield
all_tables = soup.find_all("table")
print("find_all() result:", all_tables)
```

#### ✅ What’s happening:
- We use BeautifulSoup to read the map and search for treasures:

- find("table") → finds the first table only (Luffy’s info).

- find_all("table") → finds all tables (Luffy + Asta).

 It’s like:

- Zoro finds the first pirate he sees 🗡️

- Luffy sees the whole battlefield of pirates 👀

### 🌟 Summary After 3 Cells

- First, we created an HTML page with anime data.

- Then, we stored the HTML inside Python so we can use it later.

- Finally, we used BeautifulSoup to search the HTML:

- find() → first match

- find_all() → all matches
